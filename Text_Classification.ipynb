```python
import re
def removehtml(text):
    pattern=re.compile('<.*?>')
    return pattern.sub(r'',text)
```

```python
df['review']=df['review'].apply(removehtml)
```

```python
# def punct(text):
#     if exclude in text:
#         text.remove(exclude)



import string

exclude = string.punctuation

def punct(text):
    if type(text) != str:
        return ""
    for char in exclude:
        text = text.replace(char, "")
    return text
```

```python
df['review']=df['review'].apply(punct)
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```
