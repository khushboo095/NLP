```python
import numpy as np
```

```python
df=pd.read_csv('IMDB Dataset.csv')
```

```python
df.head()
```

```python
p=df['review'].sample(100)
p
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
t=TfidfVectorizer()
```

```python
tf=t.fit_transform(p).toarray()
```

```python
tf
```

```python
print(t.idf_)
```
